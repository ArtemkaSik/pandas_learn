```python
pd.to_datetime(df['date_joined']).dt.day_of_week.unique()

m = {
    0 : 'Понедельник',
    1 : 'Вторник',
    2 : 'Среда',
    3 : 'Четверг',
    4 : 'Пятница',
    5 : 'Суббота',
    6 : 'Воскресенье'
}

df['weekday'] = pd.to_datetime(df['date_joined']).dt.day_of_week

df['weekday'] = df['weekday'].map(m)

df['date'] = pd.to_datetime(df['date_joined']).dt.date

df
```

```python
res = df.groupby('weekday')[['date']].agg(['count', 'nunique']).reset_index()

res
```

```python
import pandas as pd

df = pd.read_csv('itresume-users-pandas.csv', sep=';', skiprows=1)

fake_usernames = ['fnoxonsnbkbxsnb', 'Jonns21241412', 'fbxbnsn']

df = df[(df['id'] >= 94) & (~df['username'].isin(fake_usernames))]

df['cohort'] = pd.to_datetime(df['date_joined']).dt.strftime('%Y-%m')

res = df.groupby('cohort').size().pipe(lambda x: x / x.sum() * 100).round(2)

res.name = 'proportion'

res
```
