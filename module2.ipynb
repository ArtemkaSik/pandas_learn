```python
import pandas as pd

df = pd.read_csv('itresume-users-pandas.csv', sep=';', skiprows=1)

fake_usernames = ['fnoxonsnbkbxsnb', 'Jonns21241412', 'fbxbnsn']

df = df[(df['id'] >= 94) & (~df['username'].isin(fake_usernames))]

df['cohort'] = pd.to_datetime(df['date_joined']).dt.strftime('%Y-%m')

res = df.groupby('cohort').size().pipe(lambda x: x / x.sum() * 100).round(2)

res.name = 'proportion'

res
```

```python
import pandas as pd

df = pd.read_csv('itresume-coderun.csv', sep=',')

df
```

```python
df['cohort'] = pd.to_datetime(df['created_at']).dt.strftime('%Y-%m')

m = {
    3 : 'Python',
    2 : 'SQL'
}

df['language'] = df['language_id'].map(m)

df = df[df['cohort'] >= '2022-01']

user_tasks = df.groupby(['cohort', 'language', 'user_id'])['problem_id'].nunique().reset_index(name='task_count')

df = user_tasks.pivot_table(values='task_count', index=['cohort'], columns=['language'], aggfunc='mean')

df
```
