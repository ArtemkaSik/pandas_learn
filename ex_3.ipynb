```python
import pandas as pd
import numpy as np
```

Товары категории A:
  • Высокая оборачиваемость, большая доля выручки
  • Требуют строгого контроля, точного прогнозирования, минимальных страховых запасов

Товары категории B: 
  • Умеренный спрос, средний вклад в прибыль
  • Контроль по регламенту, периодический пересмотр запасов

Товары категории C: 
  • Низкая оборачиваемость, малая доля выручки
  • Упрощённое управление, заказ крупными партиями реже

```python
pd.set_option('display.max_columns', 40)
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
df = pd.read_csv('data/data.csv', encoding='1251')

df.head()
```

```python
groupped_df = df.groupby('DR_NDrugs').agg({'DR_Kol' : np.sum})

groupped_df
```

```python
# Вклад продажи товара в общих продажах

groupped_df = (groupped_df['DR_Kol'] / sum(groupped_df['DR_Kol'])).sort_values(ascending=False)

groupped_df
```

```python
groupped_df = groupped_df.cumsum().reset_index()

groupped_df
```

```python
groupped_df['ABC'] = np.where(groupped_df['DR_Kol'] < 0.8, 'A', np.where(groupped_df['DR_Kol'] < 0.95, 'B', 'C'))

groupped_df[groupped_df['ABC'] == 'B']
```

Многомерный ABC анализ

```python
groupped_df = df.groupby('DR_NDrugs').agg({
    'DR_Kol' : np.sum,
    'DR_CRoz' : np.sum
})

groupped_df
```

```python
groupped_df['rel_kol'] = (groupped_df['DR_Kol'] / sum(groupped_df['DR_Kol']))
groupped_df
```

```python
groupped_df = groupped_df.sort_values('rel_kol', ascending=False)
groupped_df
```

```python
groupped_df['cumsum_kol'] = groupped_df['rel_kol'].cumsum()

groupped_df
```

```python
groupped_df['ABC_kol'] = np.where(groupped_df['cumsum_kol'] < 0.8, 'A', np.where(groupped_df['cumsum_kol'] < 0.95, 'B', 'C'))

groupped_df
```
